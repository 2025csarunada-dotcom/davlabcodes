```python
# Ex-3
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the dataset
df = pd.read_csv('students.csv')

# Histogram
sns.histplot(df['Marks'], kde=True)
plt.title("Marks distribution")
plt.show()

# Box plot
sns.boxplot(x='branch', y='Marks', data=df)
plt.title("Marks by branch")
plt.show()

# Scatter plot
sns.scatterplot(x='StudyHours', y='Marks', hue='Result', data=df)
plt.title("Study hours V/S Marks")
plt.show()

# Pair plot
sns.pairplot(df[['StudyHours', 'Attendance', 'PrevMarks', 'Marks', 'Result']], hue='Result')
plt.show()
```
